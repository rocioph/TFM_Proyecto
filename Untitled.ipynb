## 2. **Carga y descripción del dataset**

#### Descripción del contenido del dataset
---
* Conjunto de datos organizado en 3 tablas parquet enlazadas:
    * videos
    * engagement_daily
    * creator_daily


#### Descripción de los campos

---

#### Tabla `videos`
* **video_id:** id. del vídeo
* **author_id:** id. del creador
* **create_time, create_date:** marca de tiempo y fecha de creación del vídeo  
* **duration:** duración del vídeo en segundos  
* **ratio:** calidad del vídeo 
* **desc_language:** idioma detectado en el texto de la descripción  
* **is_english:** indicador binario de si el idioma es en inglés  
* **desc:** descripción del vídeo escrita por el usuario  
* **sticker_text:** texto que aparece en el video de forma visual creado mediante la edicción de TikTok  
* **hashtags:** lista de objetos adjuntos al vídeo
  * {hashtags_id, hashtags_name}  
* **created_by_ai:** indicador binaria que señala si el vídeo fue creado o editado por la ia
* **is_ads:** indicador de si es contenido publicitario 
* **music_selected_from:** indica de donde viene la musica, es decir, si la musica viene de una libreria de TikTok o es un audio original creado por el usuario  
* **music_album:** nombre del album  
* **music_author:** nombre del artista, compositor o cuenta creadora del audio  
* **music_owner_id:** id. de la cuenta de TikTok propietaria del archivo de audio original  
* **music_id:** id. de la canción  
* **music_title:** titulo  
* **transcript:** transcripción del texto del audio  
* **word_count:** cantidad de palabras que contiene la descripción  
* **emoji_count:** número de emojis incluidos en el texto de la descripción  
* **question_count:** cantidad de signos de interrogación o preguntas formuladadas  
* **hashtags_count:** numero total de hashtags  
* **speaking_rate:** velocidad del audio del video  
* **gpt_summary:** resumen del video generado de forma automatica  
* **topic:** categoría a la que pertenece el video  
* **anger, joy, surprise, sadness, disgust, fear:** probabilidad de que sea esa emoción calculada a partir de la descripción 

---

#### Tabla `engagement_daily`
* **video_id:** id. del vídeo
* **date:** fecha en la que se tomaron las mediciones 
* **days_since_post:** cantidad de días transcurridos desde la fecha de publicación hasta el día en el que se tomaron las últimas medidas
* **play_count:** número de visualizaciones 
* **like_count:** número de likes/me gustas
* **comment_count:** cantidad de comentarios 
* **share_count:** número de veces que los usuarios han compartido el vídeo, tanto como fuera o dentro de la aplicación
* **collect_count:** cantidad de veces que ha sido guardado 
* **whatsapp_share_count:** número de veces que el vídeo ha sido compartido, a través de la aplicación WhatsApp

---
    
#### Tabla `creator_daily`
* **author_id:** id. del creador
* **date:** fecha en la que se registraron los ultimos datos del creador
* **follower_count:** número de seguidores
* **following_count:** cantidad de personas a las que sigue 
* **total_favorited:** total de me gusta o likes recibidos 
* **video_count:** total de videos publicados 
* **enterprise_verified:** indica si es un perfil coorporativo o de empresa verificada por TikTok (1/True) o una cuenta personal o de creador individual (0/False)


#### Tipos de emociones de la tabla "videos"
----
Probabilidad (valores del 0 al 1) de las emociones detectadas en el texto de la descripción de cada vídeo. Esto, ha sido calculada por una ia de análisis de sentimientos (NLP).

* Sentimientos analizables de cada vídeo: 
    * joy (alegria) 
    * disgust (asco)
    * sadness (tristeza)
    * anger(enfado)
    * surprise(sorpresa)
    * fear(miedo)

#### Tipos de categoria ("topic") de la tabla "video"
---
* Los diferentes temas de los que pueden ir cada video:
    * Beauty/Fashion (Belleza/Moda)
    * Others (Otros)
    * Life hacks/Personal growth (Trucos para la vida/Crecimiento personal)
    * Lifestyle (Estilo de vida)
    * Dance/Music (Danza/Música)
    * Sport/Fitness (Deporte/Fitness)
    * Cooking/Food (Cocina/Comida)
    * Shopping/Products (Compras/Producto)
    * Movies/TV/Books (Peliculas/TV/Libros)
